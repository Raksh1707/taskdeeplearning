```python
import torch
import torch.nn as nn
import time
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"

print("Device:", device)

# Input
x = torch.randn(10000, 1000, device=device)
```

```python
def swish(x, beta=1):
    return x * torch.sigmoid(beta * x)
```

```python
start = time.time()

y = swish(x)

if device == "cuda":
    torch.cuda.synchronize()

end = time.time()

# Print results
print("Input Shape :", x.shape)
print("Output Shape:", y.shape)
print("First 5 values:", y.flatten()[:5])
print("Execution Time:", round(end - start, 6), "seconds")

print("Swish activation completed successfully!")
```
