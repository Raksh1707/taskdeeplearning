```python
import torch
import torch.nn as nn
```

```python
class TCN(nn.Module):
    def __init__(self):
        super().__init__()

        self.model = nn.Sequential(
            nn.Conv1d(1, 8, 3, padding=1, dilation=1),
            nn.ReLU(),

            nn.Conv1d(8, 8, 3, padding=2, dilation=2),
            nn.ReLU(),

            nn.Conv1d(8, 8, 3, padding=4, dilation=4),
            nn.ReLU(),

            nn.Conv1d(8, 1, 3, padding=8, dilation=8)
        )

    def forward(self, x):
        return self.model(x)
```

```python
model = TCN()
```

```python
x = torch.randn(4, 1, 50)
```

```python
target = torch.randn(4, 1, 50)
```

```python
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
print(optimizer)
```

```python
for epoch in range(10):

    output = model(x)

    loss = ((output - target) ** 2).mean()

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if (epoch + 1) % 2 == 0:
        print("Epoch:", epoch + 1,
              "Loss:", round(loss.item(), 4))

print("\nInput Shape :", x.shape)
print("Output Shape:", output.shape)
print("TCN completed successfully!")
```
