```python
import torch
import torch.nn.functional as F
```

```python
# Simple CIoU Loss
def ciou_loss(pred, target):

    # IoU
    x1 = torch.max(pred[:, 0], target[:, 0])
    y1 = torch.max(pred[:, 1], target[:, 1])
    x2 = torch.min(pred[:, 2], target[:, 2])
    y2 = torch.min(pred[:, 3], target[:, 3])

    inter = torch.clamp(x2 - x1, min=0) * torch.clamp(y2 - y1, min=0)

    area1 = (pred[:, 2]-pred[:, 0]) * (pred[:, 3]-pred[:, 1])
    area2 = (target[:, 2]-target[:, 0]) * (target[:, 3]-target[:, 1])

    union = area1 + area2 - inter

    iou = inter / (union + 1e-6)

    return (1 - iou).mean()
```

```python
pred_boxes = torch.tensor([
    [10., 10., 50., 50.],
    [20., 20., 60., 60.]
])

target_boxes = torch.tensor([
    [12., 12., 48., 48.],
    [18., 18., 62., 62.]
])

# Class predictions
pred_class = torch.tensor([
    [2.0],
    [-1.0]
])

# Target classes
target_class = torch.tensor([
    [1.0],
    [0.0]
])

```

```python
box_loss = ciou_loss(pred_boxes, target_boxes)

class_loss = F.binary_cross_entropy_with_logits(
    pred_class,
    target_class
)

total_loss = box_loss + class_loss
```

```python

print("Anchor-Free Object Detection")
print("-----------------------------")
print("CIoU Loss          :", box_loss.item())
print("Classification Loss:", class_loss.item())
print("Total Loss         :", total_loss.item())
```
