```python
import torch
import numpy as np
```

```python
x = torch.randn(4, 5)
x.requires_grad = False

gamma = torch.ones(5)
beta = torch.zeros(5)
```

```python

# Batch Normalization
def batch_norm(x):

    mean = x.mean(dim=0)
    var = x.var(dim=0, unbiased=False)

    x_hat = (x - mean) / torch.sqrt(var + 1e-5)

    out = gamma * x_hat + beta

    return out, mean, var, x_hat
```

```python
def layer_norm(x):

    mean = x.mean(dim=1, keepdim=True)
    var = x.var(dim=1, keepdim=True, unbiased=False)

    x_hat = (x - mean) / torch.sqrt(var + 1e-5)

    out = gamma * x_hat + beta

    return out, x_hat
```

```python
 # Manual Backward
# -----------------------
def backward(dout, x_hat):

    dgamma = torch.sum(dout * x_hat, dim=0)
    dbeta = torch.sum(dout, dim=0)

    return dgamma, dbeta

# BatchNorm
bn_out, running_mean, running_var, bn_hat = batch_norm(x)

# LayerNorm
ln_out, ln_hat = layer_norm(x)

# Dummy Gradient
dout = torch.ones_like(x)

# Backward
dgamma_bn, dbeta_bn = backward(dout, bn_hat)
dgamma_ln, dbeta_ln = backward(dout, ln_hat)
```

```python
print("BatchNorm Output")
print(bn_out)

print("\nLayerNorm Output")
print(ln_out)

print("\nRunning Mean")
print(running_mean)

print("\nRunning Variance")
print(running_var)

print("\nBatchNorm dGamma:", dgamma_bn)
print("BatchNorm dBeta :", dbeta_bn)

print("\nLayerNorm dGamma:", dgamma_ln)
print("LayerNorm dBeta :", dbeta_ln)
```
