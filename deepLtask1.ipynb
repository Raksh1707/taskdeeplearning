```python
import numpy as np

# For same output every run
np.random.seed(42)
```

```python
# X -> Input Features
# y -> Target Class
X = np.array([
    [0.2,0.5],
    [0.8,0.3],
    [0.4,0.9],
    [0.7,0.6]
])

y = np.array([0,1,1,0])
```

```python
num_classes = 2

# Convert labels into One-Hot Encoding
Y = np.eye(num_classes)[y]
```

```python
 #Input -> Hidden -> Output

input_size = 2
hidden_size = 4
output_size = 2
```

```python
#Initialize Weights
# Small random values improve learning

W1 = np.random.randn(input_size, hidden_size) * 0.1
b1 = np.zeros((1, hidden_size))

W2 = np.random.randn(hidden_size, output_size) * 0.1
b2 = np.zeros((1, output_size))
```

```python
# Learning Rate
# Controls weight update size
learning_rate = 0.1
```

```python
# ReLU Activation
# Removes negative values

def relu(x):
    return np.maximum(0, x)

# Gradient of ReLU
def relu_derivative(x):
    return (x > 0).astype(float)
```

```python
# Converts scores into probabilities


def softmax(x):

    # Improve numerical stability
    exp = np.exp(x - np.max(x, axis=1, keepdims=True))

    return exp / np.sum(exp, axis=1, keepdims=True)
```

```python
# Cross Entropy Loss
# Measures prediction error

def cross_entropy(y_true, y_pred):

    m = y_true.shape[0]

    loss = -np.sum(y_true * np.log(y_pred + 1e-8)) / m

    return loss
```

```python
epochs = 1000

for epoch in range(epochs):

    # Forward Pass
    Z1 = X @ W1 + b1
    A1 = relu(Z1)

    Z2 = A1 @ W2 + b2
    A2 = softmax(Z2)

    loss = cross_entropy(Y, A2)

    # Backward Pass
    m = X.shape[0]

    # Output layer error
    dZ2 = (A2 - Y) / m

    # Output gradients
    dW2 = A1.T @ dZ2
    db2 = np.sum(dZ2, axis=0, keepdims=True)

    # Hidden layer error
    dA1 = dZ2 @ W2.T
    dZ1 = dA1 * relu_derivative(Z1)

    # Hidden gradients
    dW1 = X.T @ dZ1
    db1 = np.sum(dZ1, axis=0, keepdims=True)

    # Update weights
    W1 -= learning_rate * dW1
    b1 -= learning_rate * db1

    W2 -= learning_rate * dW2
    b2 -= learning_rate * db2

    if epoch % 100 == 0:
        print(f"Epoch {epoch}, Loss = {loss:.4f}")
```

```python

print("\nFinal Prediction Probabilities:")
print(A2)

prediction = np.argmax(A2, axis=1)

print("\nPredicted Classes:")
print(prediction)

print("\nActual Classes:")
print(y)

accuracy = np.mean(prediction == y) * 100

print(f"\nAccuracy = {accuracy:.2f}%")
```
