```python
import torch
```

```python
input_size = 4
hidden_size = 3
seq_len = 5
batch_size = 2

# Input
x = torch.randn(batch_size, seq_len, input_size)
```

```python
def lstm(x, hidden_size):
    h = torch.zeros(x.size(0), hidden_size)
    c = torch.zeros(x.size(0), hidden_size)

    W = torch.randn(input_size + hidden_size, 4 * hidden_size)

    outputs = []

    for t in range(x.size(1)):
        z = torch.cat((x[:, t, :], h), dim=1)

        gates = z @ W

        i, f, o, g = torch.chunk(gates, 4, dim=1)

        i = torch.sigmoid(i)
        f = torch.sigmoid(f)
        o = torch.sigmoid(o)
        g = torch.tanh(g)

        c = f * c + i * g
        h = o * torch.tanh(c)

        outputs.append(h)

    return torch.stack(outputs, dim=1)
```

```python
forward = lstm(x, hidden_size)

# Backward LSTM
backward = lstm(torch.flip(x, [1]), hidden_size)
backward = torch.flip(backward, [1])

# BiLSTM output
output = torch.cat((forward, backward), dim=2)
```
