```python
import torch
import torch.nn as nn
```

```python
class Autoencoder(nn.Module):
    def __init__(self):
        super().__init__()

        self.encoder = nn.Sequential(
            nn.Conv2d(1, 8, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2)
        )

        self.decoder = nn.Sequential(
            nn.ConvTranspose2d(8, 1, 2, stride=2),
            nn.Sigmoid()
        )

    def forward(self, x):
        return self.decoder(self.encoder(x))
```

```python
model = Autoencoder()

optimizer = torch.optim.Adam(model.parameters(), lr=0.001)

# Normal images
images = torch.rand(10, 1, 28, 28)

# Training
for epoch in range(10):

    output = model(images)

    loss = ((images - output) ** 2).mean()

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if (epoch + 1) % 2 == 0:
        print("Epoch:", epoch + 1, "Loss:", round(loss.item(), 4))
```

```python
test = torch.rand(1, 1, 28, 28)

with torch.no_grad():
    result = model(test)

error = ((test - result) ** 2).mean().item()

print("\nAnomaly Score:", round(error, 4))

if error > 0.1:
    print("Anomaly Detected")
else:
    print("Normal Image")

print("Autoencoder completed successfully!")
```
